# Part 2: BeautifulSoup Web Scraping
```
The company wants some additional external information about major cities or locations.

Using:
● requests
● BeautifulSoup

scrape a public webpage containing city-related information.
Collect approximately 10 to 20 records.

Example format:
City Population or Rank
Mumbai Value
Delhi Value
Bengaluru Value
Your objective is only to demonstrate basic web scraping.

Your program should contain:
● requests.get()
● BeautifulSoup object creation
● find() or find_all()
● Loop through webpage elements
● Store collected information inside a list
● Convert the information into a Pandas DataFrame

Export the scraped data as:
external_city_data.csv
```
___________

````python
import requests
from bs4 import BeautifulSoup
import pandas as pd
````

### Webpage URL

````python
url = "https://statisticstimes.com/demographics/city/india-cities-population.php"
````

### Request

````python
response = requests.get(url)
print("Status Code:", response.status_code)
````

### Create BeautifulSoup object


````python
soup = BeautifulSoup(response.text, "html.parser")
````

### Find all table rows

````python
rows = soup.find_all("tr")
````

### Empty list to store scraped data


````python
city_data = []
````

### Loop through rows


````python
for row in rows:
    columns = row.find_all("td")

    if len(columns) >= 4:
        city = columns[0].get_text(strip=True)
        population_2025 = columns[1].get_text(strip=True)
        population_2026 = columns[2].get_text(strip=True)
        rank = columns[4].get_text(strip=True)

        city_data.append({
            "City": city,
            "Population_2025": population_2025,
            "Population_2026": population_2026,
            "Rank": rank
        })
````

### Take only first 20 records


````python
city_data = city_data[:20]
city_data
````

### Convert list into DataFrame


````python
df = pd.DataFrame(city_data)
df
````

### Display the data


````python
print("\nScraped City Data:")
print(df)
````

### Export to CSV

````python
df.to_csv("external_city_data.csv", index=False)
print("\nData successfully exported to external_city_data.csv")
````
