### **Check missing values**
- using `.isnull()` and `.sum()`
- `isnull()` using to check and view null values
- `sum()` using to calculate the count of missing values.


````python
print("Missing Values\n")
delivery_details.isnull()
````

## Working of `.isnull()`
- if the cell has **`NO DATA`** then it will show **`TRUE`**
- if the cell has **`DATA`** then it will show **`FALSE`**
__________________________________________________________________________________________________________
## Working of `.isnull().sum()`
- `.sum()` using to count the total `null values`
- if `0` ,indicates column has `NO NULL VALUES` in column
- if `COUNT OF NULL VALUES` ,incicates column has `NULL VALUES` in column
____________________________________________________________________________________________________________
- **What it does:** It adds up the results from the step above for each column.
- **Behind the scenes:** In Python, the computer reads True as 1 and False as 0.
-  By summing them up, Python is literally counting how many True (missing) values exist in each column.

````python
print("Missing Values\n")
delivery_details.isnull().sum()
````

````python
median_rating = delivery_details["customer_rating"].median()

delivery_details["customer_rating"] = delivery_details["customer_rating"].fillna(median_rating)
````

````python
delivery_details.isnull().sum()
````

````python
median_duration = delivery_details["delivery_duration_min"].median()

delivery_details["delivery_duration_min"] = delivery_details[
    "delivery_duration_min"
].fillna(median_duration)
````

````python
delivery_details.isnull().sum()
````

````python
delivery_details["delivery_time"] = delivery_details["delivery_time"].fillna(
    delivery_details["order_time"]
)
````

````python
delivery_details.isnull().sum()
````

````python
delivery_details
````

# Step 3: Create a New Column
```
Create a column called:
Delivery_Status

Use the following conditions:
● Delivery time less than or equal to 30 minutes: Fast
● Delivery time between 31 and 45 minutes: Normal
● Delivery time greater than 45 minutes: Delayed
```

````python
import numpy as np
````

````python
delivery_details["Delivery_Status"] = np.where(
    delivery_details["delivery_duration_min"] <= 30,
    "Fast",
    np.where(
        delivery_details["delivery_duration_min"] <= 45,
        "Normal",
        "Delayed"
    )
)
````

````python
delivery_details.head()
````

# Step 4: Perform Basic Analysis
```
Using Pandas, calculate:
1. Total number of orders
2. Average delivery time
3. Cancellation percentage
4. Average customer rating
5. Restaurant with the highest average delivery time
6. Location with the highest number of delayed orders

Export the cleaned dataset as:
cleaned_food_delivery.csv
```

### Total number of orders

````python
total_orders = delivery_details["order_id"].count()
print("Total number of orders:", total_orders)
````

### Average delivery time 


````python
avg_delivery_time = delivery_details["delivery_duration_min"].mean()
print("Average delivery time:", round(avg_delivery_time, 2))
````

### Cancellation percentage


````python
cancellation_percentage = delivery_details["is_canceled"].mean() * 100
print("Cancellation Percentage:", round(cancellation_percentage, 2), "%")
````

### Average customer rating

````python
average_rating = delivery_details["customer_rating"].mean()
print("Average Customer Rating:", round(average_rating, 2))
````

###  Restaurant with the highest average delivery time


````python
restaurant_avg_delivery = delivery_details.groupby("restaurant")["delivery_duration_min"].mean()
restaurant_highest_delivery = restaurant_avg_delivery.idxmax()
print("Restaurant:", restaurant_highest_delivery)
print("Average Delivery Time:", restaurant_avg_delivery.max())
````

### Location with the highest number of delayed orders


````python
delayed_orders = delivery_details[delivery_details["Delivery_Status"] == "Delayed"]
delayed_by_location = delayed_orders.groupby("city").size()
print("Location with highest delayed orders:", delayed_by_location.idxmax())
print("Number of delayed orders:", delayed_by_location.max())
````
